````python

### 1. Questões Teóricas

* **1. (TEÓRICA) Encapsulamento e acesso a dados**
  * **Resposta Correta: C**
  * **Explicação:** Em Python, o prefixo com duplo *underscore* (`__idade`) ativa o mecanismo de *name mangling*, dificultando o acesso direto de fora da classe e indicando que o atributo só deve ser acessado ou modificado por meio de métodos internos (getters e setters),.

* **2. (TEÓRICA) Pilares da POO**
  * **Resposta Correta: D**
  * **Explicação:** Trata-se do **Polimorfismo**, que permite que objetos de diferentes classes (`Carro` e `Barco`) respondam ao mesmo método (`mover()`) com comportamentos específicos de cada tipo.

* **3. (TEÓRICA) Comparando paradigmas**
  * **Resposta Correta: B**
  * **Explicação:** A orientação a objetos organiza a aplicação agrupando estado (atributos) e comportamento (métodos) na mesma estrutura (classes/objetos), ao passo que a programação procedural separa as funções das estruturas de dados.

* **4. (TEÓRICA) Conceitos de Python**
  * **Resposta Correta: C**
  * **Explicação:** A alternativa C está **incorreta** porque o Python **suporta herança múltipla**, permitindo que uma subclasse herde diretamente de duas ou mais superclasses.

---

### 2. Exercícios Práticos

#### Exercício 5: Classe Pessoa
```python
class Pessoa:

  def __init__(self, nome: str, idade: int):
    self.nome = nome
    self.idade = idade

  def faz_aniversario(self):
    self.idade += 1

  def __str__(self):
    return f"{self.nome}, {self.idade} anos"

# Teste do Exercício 5
p1 = Pessoa("Ana", 25)
p2 = Pessoa("Carlos", 30)

print("--- Antes do aniversário ---")
print(p1)
print(p2)

p1.faz_aniversario()
p2.faz_aniversario()

print("\n--- Depois do aniversário ---")
print(p1)
print(p2)
```

---

#### Exercício 6: Encapsulamento na Classe Lampada,
```python
class Lampada:

  def __init__(self):
    self.__estado = False  # Atributo privado

  def ligar(self):
    self.__estado = True

  def desligar(self):
    self.__estado = False

  def is_acesa(self) -> bool:
    return self.__estado

# Teste do Exercício 6
lampada = Lampada()
print("Lâmpada acesa?", lampada.is_acesa())  # False

lampada.ligar()
print("Lâmpada acesa?", lampada.is_acesa())  # True

lampada.desligar()
print("Lâmpada acesa?", lampada.is_acesa())  # False
```

---

#### Exercício 7: Herança (Animal, Cachorro e Gato)
```python
class Animal:

  def fazer_som(self):
    print("Algum som")

class Cachorro(Animal):

  def fazer_som(self):
    print("Au Au")

class Gato(Animal):

  def fazer_som(self):
    print("Miau")

# Teste do Exercício 7
animal_generico = Animal()
dog = Cachorro()
cat = Gato()

animal_generico.fazer_som()  # Algum som
dog.fazer_som()  # Au Au
cat.fazer_som()  # Miau
```

---

#### Exercício 8: Polimorfismo com `animais_falam`
```python
def animais_falam(lista_animais):
  for animal in lista_animais:
    animal.fazer_som()

# Teste do Exercício 8
meus_animais = [Cachorro(), Gato(), Cachorro(), Gato()]
animais_falam(meus_animais)
```

---

### **3. Projetos**

#### **Projeto 1: Biblioteca de Livros,**
```python
class Livro:

  def __init__(self, titulo: str, autor: str, quantidade: int):
    self.titulo = titulo
    self.autor = autor
    self.__qtd_disponivel = quantidade  # Atributo protegido contra modificação direta

  def get_qtd_disponivel(self):
    return self.__qtd_disponivel

  def emprestar() -> bool:
    if self.__qtd_disponivel > 0:
      self.__qtd_disponivel -= 1
      return True
    return False

  def devolver():
    self.__qtd_disponivel += 1

  def __str__(self):
    return f"'{self.titulo}' por {self.autor} (Disponíveis: {self.__qtd_disponivel})"

class Usuario:

  def __init__(self, nome: str, matricula: str):
    self.nome = nome
    self.matricula = matricula
    self.livros_emprestados = []

  def pegar_emprestado(self, livro: Livro):
    if livro.emprestar():
      self.livros_emprestados.append(livro)
      print(f"{self.nome} pegou emprestado o livro '{livro.titulo}'.")
    else:
      print(f"Não foi possível emprestar '{livro.titulo}': sem estoque.")

  def devolver_livro(self, livro: Livro):
    if livro in self.livros_emprestados:
      livro.devolver()
      self.livros_emprestados.remove(livro)
      print(f"{self.nome} devolveu o livro '{livro.titulo}'.")
    else:
      print(f"{self.nome} não possui o livro '{livro.titulo}' em mãos.")

# Teste do Projeto 1
livro1 = Livro("Estruturas de Dados", "Ana Silva", 1)
user = Usuario("João", "2026001")

print(livro1)
user.pegar_emprestado(livro1)  # Sucesso
user.pegar_emprestado(livro1)  # Indisponível
user.devolver_livro(livro1)  # Devolvido
print(livro1)
```

---

#### **Projeto 2: Gerenciamento de Conta Bancária com Herança,,**
```python
class Conta:

  def __init__(self, titular: str, saldo_inicial: float = 0.0):
    self.titular = titular
    self._saldo = saldo_inicial  # Encapsulado para controle de acesso

  def depositar(self, valor: float):
    if valor > 0:
      self._saldo += valor
      print(
          f"Depósito de R${valor:.2f} realizado na conta de {self.titular}."
      )

  def sacar(self, valor: float) -> bool:
    if 0 < valor <= self._saldo:
      self._saldo -= valor
      print(f"Saque de R${valor:.2f} realizado com sucesso.")
      return True
    print(f"Saldo insuficiente para saque de R${valor:.2f}.")
    return False

  def get_saldo(self) -> float:
    return self._saldo

class ContaCorrente(Conta):

  def __init__(
      self,
      titular: str,
      saldo_inicial: float = 0.0,
      taxa_saque: float = 2.50,
  ):
    super().__init__(titular, saldo_inicial)
    self.taxa_saque = taxa_saque

  def sacar(self, valor: float) -> bool:
    valor_total = valor + self.taxa_saque
    if 0 < valor_total <= self._saldo:
      self._saldo -= valor_total
      print(
          f"Saque de R${valor:.2f} (Taxa: R${self.taxa_saque:.2f}) efetuado."
      )
      return True
    print("Saldo insuficiente cobrindo a taxa de saque.")
    return False

class ContaPoupanca(Conta):

  taxa_juros = 0.05  # Atributo de classe (5% ao ano)

  def __init__(self, titular: str, saldo_inicial: float = 0.0):
    super().__init__(titular, saldo_inicial)

  def render_juros(self):
    rendimento = self._saldo * ContaPoupanca.taxa_juros
    self._saldo += rendimento
    print(
        f"Juros de R${rendimento:.2f} aplicados. Novo saldo:"
        f" R${self._saldo:.2f}"
    )

# Teste do Projeto 2
cc = ContaCorrente("Maria", 100.0)
cp = ContaPoupanca("José", 500.0)

cc.sacar(20.0)  # Deduz 20.0 + 2.50
print("Saldo Conta Corrente:", cc.get_saldo())

cp.render_juros()  # Rende 5% sobre 500
print("Saldo Conta Poupança:", cp.get_saldo())
```
````
