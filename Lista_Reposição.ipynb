````python


---

### **Parte A — Questões de Múltipla Escolha**

* **Exercício 1 — Objetos, encapsulamento e interface**
  * **Resposta Correta: B**
  * *Explicação:* O encapsulamento permite ocultar a representação interna do saldo (`__saldo`) e disponibilizar uma interface de alto nível (métodos públicos `depositar`, `sacar` e `consultar_saldo`) para a manipulação do objeto.

* **Exercício 2 — Herança e polimorfismo**
  * **Resposta Correta: C**
  * *Explicação:* As classes `Vendedor` e `Gerente` herdam de `Funcionario` (Herança) e redefinem o método `calcular_bonus()` (Polimorfismo), permitindo que cada classe responda de maneira adequada ao mesmo método.

* **Exercício 3 — Atributo de instância x atributo de classe**
  * **Resposta Correta: B**
  * *Explicação:* O atributo `contador` pertence à classe `Produto` e é compartilhado entre todas as suas instâncias, acumulando o valor 3 após as três instanciações.

---

### **Parte B — Programação: Nível Fácil**

#### **Exercício 4 — Classe Livro: atributos e métodos**
```python
class Livro:

  def __init__(self, titulo: str, autor: str, paginas: int):
    self.titulo = titulo
    self.autor = autor
    self.paginas = paginas

  def descricao(self) -> str:
    return f"{self.titulo} — {self.autor} ({self.paginas} páginas)"

# Programa principal
livro1 = Livro("Dom Casmurro", "Machado de Assis", 256)
livro2 = Livro("O Hobbit", "J.R.R. Tolkien", 310)

print(livro1.descricao())
print(livro2.descricao())
```

#### **Exercício 5 — Conta simples com encapsulamento**
```python
class Conta:

  def __init__(self, titular: str):
    self.titular = titular
    self.__saldo = 0.0  # Atributo privado

  def depositar(self, valor: float):
    if valor > 0:
      self.__saldo += valor  # Permite apenas valores positivos

  def consultar_saldo(self) -> float:
    return self.__saldo

# Programa principal
minha_conta = Conta("Carlos")
minha_conta.depositar(150.0)
minha_conta.depositar(250.50)
print(
    f"Titular: {minha_conta.titular} | Saldo atual: R$"
    f" {minha_conta.consultar_saldo():.2f}"
)
```

#### **Exercício 6 — Animais: herança e polimorfismo**
```python
class Animal:

  def __init__(self, nome: str):
    self.nome = nome

  def emitir_som(self) -> str:
    return "Som genérico"

class Cachorro(Animal):

  def emitir_som(self) -> str:
    return "au au"

class Gato(Animal):

  def emitir_som(self) -> str:
    return "miau"

# Programa principal
animais = [Cachorro("Rex"), Gato("Mingau")]

for animal in animais:
  # Executa emitir_som() de forma polimórfica sem checar o tipo
  print(f"{animal.nome} faz: {animal.emitir_som()}")
```

---

### **Parte C — Programação: Nível Médio**

#### **Exercício 7 — Veículo com velocidade controlada**
```python
class Veiculo:

  def __init__(self, placa: str, velocidade_max: float):
    self.__placa = placa
    self.__velocidade_max = velocidade_max
    self.__velocidade_atual = 0.0  # Inicia em 0

  def get_placa(self) -> str:
    return self.__placa

  def get_velocidade_max(self) -> float:
    return self.__velocidade_max

  def set_velocidade_max(self, nova_max: float):
    # Aceita apenas valores positivos e nunca menor que a velocidade atual
    if nova_max > 0 and nova_max >= self.__velocidade_atual:
      self.__velocidade_max = nova_max

  def acelerar(self, incremento: float = 10):
    if incremento > 0:
      self.__velocidade_atual = min(
          self.__velocidade_max, self.__velocidade_atual + incremento
      )  # Não ultrapassa a máxima

  def frear(self, decremento: float = 10):
    if decremento > 0:
      self.__velocidade_atual = max(
          0.0, self.__velocidade_atual - decremento
      )  # Não fica negativa

  def __str__(self) -> str:
    return (
        f"Veículo [{self.__placa}] - Vel. Atual: {self.__velocidade_atual} km/h"
        f" | Máx: {self.__velocidade_max} km/h"
    )

# Programa principal
carro = Veiculo("ABC-1234", 100.0)
print(carro)

carro.acelerar(50)
carro.acelerar(70)  # Deve limitar em 100 km/h
print(carro)

carro.frear(30)
print(carro)

carro.frear(100)  # Deve limitar em 0 km/h
print(carro)
```

#### **Exercício 8 — Funcionários e cálculo de bônus**
```python
class Funcionario:

  def __init__(self, nome: str, salario: float):
    self.nome = nome
    self.salario = salario

  def calcular_bonus(self) -> float:
    return self.salario * 0.05  # 5% do salário

  def __str__(self) -> str:
    return f"{self.nome} - Salário: R$ {self.salario:.2f}"

class Vendedor(Funcionario):

  def __init__(self, nome: str, salario: float, vendas: float):
    super().__init__(nome, salario)
    self.vendas = vendas

  def calcular_bonus(self) -> float:
    return (self.salario * 0.05) + (self.vendas * 0.02)  # 5% sal + 2% vendas

class Gerente(Funcionario):

  def __init__(self, nome: str, salario: float, equipe: list):
    super().__init__(nome, salario)
    self.equipe = equipe

  def calcular_bonus(self) -> float:
    return (self.salario * 0.10) + (
        100.0 * len(self.equipe)
    )  # 10% sal + R$100 p/ integrante

# Programa principal
f1 = Funcionario("Ana", 3000.0)
v1 = Vendedor("Bruno", 2000.0, 50000.0)
g1 = Gerente("Carla", 8000.0, ["Ana", "Bruno"])

equipe_empresa = [f1, v1, g1]

for f in equipe_empresa:
  print(
      f"Nome: {f.nome} | Tipo: {type(f).__name__} | Bônus: R$"
      f" {f.calcular_bonus():.2f}"
  )
```

#### **Exercício 9 — Produtos e contador de objetos**
```python
class Produto:

  contador = 0  # Atributo de classe

  def __init__(self, nome: str, preco: float):
    Produto.contador += 1
    self.__id = Produto.contador  # Id sequencial automático
    self.__nome = nome
    self.__preco = preco

  def get_id(self) -> int:
    return self.__id

  def get_nome(self) -> str:
    return self.__nome

  def get_preco(self) -> float:
    return self.__preco

  def set_preco(self, novo_preco: float):
    if novo_preco > 0:  # Apenas valores maiores que zero
      self.__preco = novo_preco

  def aplicar_desconto(self, percentual: float):
    if 0 <= percentual <= 100:  # Validação entre 0 e 100
      self.__preco -= self.__preco * (percentual / 100.0)

  def __str__(self) -> str:
    return f"ID {self.__id} | {self.__nome} - R$ {self.__preco:.2f}"

# Programa principal
p1 = Produto("Teclado", 150.0)
p2 = Produto("Mouse", 80.0)
p3 = Produto("Monitor", 900.0)

p1.aplicar_desconto(10)  # 10% de desconto no Teclado

print(p1)
print(p2)
print(p3)
print(f"Total de produtos criados: {Produto.contador}")  # Exibe 3
```

---

### **Parte D — Desafio Integrador**

#### **Exercício 10 — Mini sistema de pedidos de e-commerce**
```python
class Produto:

  contador = 0  # Atributo de classe para ID automático

  def __init__(self, nome: str, preco: float):
    Produto.contador += 1
    self.__id = Produto.contador
    self.__nome = nome
    self.__preco = preco

  def get_id(self) -> int:
    return self.__id

  def get_nome(self) -> str:
    return self.__nome

  def get_preco(self) -> float:
    return self.__preco

  def set_preco(self, novo_preco: float):
    if novo_preco > 0:
      self.__preco = novo_preco

  def calcular_frete(self) -> float:
    return 0.0  # Retorna 0 por padrão

  def __str__(self) -> str:
    return f"[{self.__id}] {self.__nome} - R$ {self.__preco:.2f}"

class ProdutoFisico(Produto):

  def __init__(self, nome: str, preco: float, peso: float):
    super().__init__(nome, preco)
    self.peso = peso

  def calcular_frete(self) -> float:
    return 5.0 + (2.0 * self.peso)  # R$ 5,00 + R$ 2,00 por kg

class ProdutoDigital(Produto):

  def __init__(self, nome: str, preco: float, tamanho_mb: float):
    super().__init__(nome, preco)
    self.tamanho_mb = tamanho_mb

  def calcular_frete(self) -> float:
    return 0.0  # Frete igual a 0

class Cliente:

  def __init__(self, nome: str, email: str):
    self.__nome = nome
    self.__email = email

  def get_nome(self) -> str:
    return self.__nome

  def get_email(self) -> str:
    return self.__email

  def __str__(self) -> str:
    return f"Cliente: {self.__nome} ({self.__email})"

class Pedido:

  contador_pedidos = 0  # Atributo de classe para número automático

  def __init__(self, cliente: Cliente):
    Pedido.contador_pedidos += 1
    self.numero = Pedido.contador_pedidos
    self.cliente = cliente
    self.produtos = []
    self.__status = "ABERTO"  # Status privado

  def adicionar_produto(self, produto: Produto):
    if self.__status == "ABERTO":
      self.produtos.append(produto)
      print(f"Produto '{produto.get_nome()}' adicionado ao pedido #{self.numero}.")
    else:
      print(
          f"Erro: O pedido #{self.numero} está fechado e não aceita novos"
          " produtos."
      )  # Regra de controle interno

  def calcular_total(self) -> float:
    return sum(p.get_preco() for p in self.produtos)  # Soma preços dos produtos

  def calcular_frete_total(self) -> float:
    # Polimorfismo: chama calcular_frete() sem checar tipo
    return sum(p.calcular_frete() for p in self.produtos)

  def fechar_pedido(self):
    self.__status = "FECHADO"
    print(f"Pedido #{self.numero} fechado com sucesso.")

  def __str__(self) -> str:
    resumo_produtos = "\n".join([f"  - {p}" for p in self.produtos])
    total = self.calcular_total()
    frete = self.calcular_frete_total()
    return (
        f"--- Pedido #{self.numero} [{self.__status}] ---\n"
        f"{self.cliente}\n"
        f"Itens:\n{resumo_produtos}\n"
        f"Subtotal: R$ {total:.2f} | Frete Total: R$ {frete:.2f} | Total Geral:"
        f" R$ {(total + frete):.2f}\n"
        "-------------------------------------"
    )

# Programa Principal (Demonstração)
if __name__ == "__main__":
  # 1. Criar cliente
  cliente1 = Cliente("Lucas Mendes", "lucas@email.com")

  # 2. Criar produtos físicos e digitais
  livro_fisico = ProdutoFisico("Livro de Python", 80.0, peso=1.5)  # Frete: 5 + 2*1.5 = R$ 8.00
  e_book = ProdutoDigital("E-book de POO", 30.0, tamanho_mb=15.0)  # Frete: R$ 0.00
  cadeira = ProdutoFisico("Cadeira Gamer", 1200.0, peso=15.0)       # Frete: 5 + 2*15 = R$ 35.00

  # 3. Montar pedido e adicionar produtos
  pedido1 = Pedido(cliente1)
  pedido1.adicionar_produto(livro_fisico)
  pedido1.adicionar_produto(e_book)
  pedido1.adicionar_produto(cadeira)

  # 4. Exibir resumo
  print("\n" + str(pedido1))

  # 5. Fechar pedido
  pedido1.fechar_pedido()

  # 6. Tentar adicionar mais um produto para comprovar a regra de bloqueio
  curso_online = ProdutoDigital("Curso em Vídeo", 100.0, tamanho_mb=500.0)
  pedido1.adicionar_produto(curso_online)
```
````
