## Exercício 1
**Resposta: O(1).** O acesso ao primeiro elemento é feito diretamente e não depende do tamanho da lista.

```python
def imprime_primeiro(lista):
    print(lista[0])

# Complexidade: O(1)
```

## Exercício 2
**Resposta: O(n).** A função percorre a lista para encontrar o maior valor.

```python
def encontrar_maximo(lista):
    valor_maximo = lista[0]
    for numero in lista:
        if numero > valor_maximo:
            valor_maximo = numero
    return valor_maximo

# Complexidade: O(n)
```

## Exercício 3
**Resposta: O(log n).** A busca binária reduz o espaço de busca aproximadamente pela metade a cada iteração.

```python
def busca_binaria(lista, alvo):
    esquerda = 0
    direita = len(lista) - 1
    while esquerda <= direita:
        meio = (esquerda + direita) // 2
        if lista[meio] == alvo:
            return meio
        elif lista[meio] < alvo:
            esquerda = meio + 1
        else:
            direita = meio - 1
    return -1

# Complexidade: O(log n)
```

## Exercício 5
**Resposta: O(2^n).** A implementação recursiva faz várias chamadas repetidas, formando uma árvore de chamadas de crescimento exponencial.

```python
def fib(n):
    if n < 2:
        return n
    return fib(n-1) + fib(n-2)

# Complexidade de tempo: O(2^n)
```
