````python
### Respostas das Questões

#### **1. Resposta da Questão 1 (ENADE)**
**Alternativa correta: B**.

* **Justificativa:** Na **busca binária**, o melhor caso ocorre quando o elemento buscado está localizado exatamente no ponto médio inicial, resultando em complexidade **\\(O(1)\\)**. No pior caso (quando o elemento não está no vetor ou se encontra nas extremidades), dividimos o vetor sucessivamente ao meio, resultando em complexidade **\\(O(\log n)\\)**. Na **busca sequencial**, o melhor caso ocorre ao encontrar o elemento na primeira posição (**\\(O(1)\\)**), enquanto o pior caso exige a verificação de todos os \\(n\\) elementos (**\\(O(n)\\)**).

---

#### **2. Resposta da Questão 2 (Teórica)**
* **Por que a ordenação é exigida:** O algoritmo de **busca binária** fundamenta sua lógica na ordenação do vetor para poder descartar metade dos elementos a cada iteração. Ao comparar o valor buscado com o valor na posição central (`vetor[meio]`), ele decide se deve prosseguir a busca no subvetor à esquerda (se o alvo for menor) ou no subvetor à direita (se o alvo for maior).
* **O que aconteceria se o conjunto não estivesse ordenado:** Se o vetor não estiver ordenado, a premissa do algoritmo falha. Ao descartar uma das metades, o algoritmo poderá eliminar justamente o segmento onde o elemento procurado estava. Isso resultará em **falsos negativos** (não encontrar um elemento existente no vetor) ou em índices de retorno incorretos.

---

### Código em Python (`capitulo04_exercicios.py`)

Abaixo está o script completo em Python contendo as resoluções das questões teóricas em formato de comentários e a implementação prática das questões 3 e 4. Você pode copiar o código abaixo e salvá-lo como um arquivo `.py`.

```python
# ==============================================================================
# CAPÍTULO 4: VETORES ORDENADOS E BUSCA BINÁRIA
# Arquivo Python contendo as resoluções das questões do capítulo
# ==============================================================================

# ------------------------------------------------------------------------------
# EXERCÍCIO 1 (ENADE)
# ------------------------------------------------------------------------------
# Resposta correta: B
# Melhor caso O(1) e pior caso O(log n) na busca binária, enquanto a busca
# sequencial tem melhor caso O(1) e pior caso O(n).

# ------------------------------------------------------------------------------
# EXERCÍCIO 2 (TEÓRICA)
# ------------------------------------------------------------------------------
# A busca binária exige que o vetor esteja ordenado porque ela decide descartar
# metade dos elementos comparando o alvo com o elemento do meio. Se o vetor
# não estiver ordenado, o elemento pode estar na metade descartada, gerando
# respostas incorretas.

# ------------------------------------------------------------------------------
# EXERCÍCIO 3 (PRÁTICA): Busca Binária Recursiva
# ------------------------------------------------------------------------------
def busca_binaria(vetor: list, alvo: int, esq: int = 0, dir: int = None) -> int:
    """
    Realiza a busca binária de forma recursiva em um vetor ordenado.

    Parâmetros:
        vetor (list): Lista de elementos ordenados.
        alvo (int): Valor a ser buscado.
        esq (int): Índice esquerdo do limite de busca.
        dir (int): Índice direito do limite de busca.

    Retorna:
        int: O índice do alvo no vetor se encontrado, ou -1 caso contrário.
    """
    if dir is None:
        dir = len(vetor) - 1

    # Caso base: segmento inválido (elemento não está no vetor)
    if esq > dir:
        return -1

    meio = (esq + dir) // 2

    # Caso base: elemento encontrado
    if vetor[meio] == alvo:
        return meio
    # Passos recursivos
    elif vetor[meio] > alvo:
        return busca_binaria(vetor, alvo, esq, meio - 1)
    else:
        return busca_binaria(vetor, alvo, meio + 1, dir)

# ------------------------------------------------------------------------------
# EXERCÍCIO 4 (PRÁTICA): Comparação de Comparações (Pior Caso)
# ------------------------------------------------------------------------------
def busca_sequencial_com_contagem(vetor: list, alvo: int) -> tuple[int, int]:
    """Realiza a busca sequencial e retorna (índice, total_de_comparacoes)."""
    comparacoes = 0
    for i in range(len(vetor)):
        comparacoes += 1
        if vetor[i] == alvo:
            return i, comparacoes
    return -1, comparacoes

def busca_binaria_com_contagem(vetor: list, alvo: int) -> tuple[int, int]:
    """Realiza a busca binária iterativa e retorna (índice, total_de_comparacoes)."""
    esq, dir = 0, len(vetor) - 1
    comparacoes = 0

    while esq <= dir:
        meio = (esq + dir) // 2
        comparacoes += 1

        if vetor[meio] == alvo:
            return meio, comparacoes
        elif vetor[meio] < alvo:
            esq = meio + 1
        else:
            dir = meio - 1

    return -1, comparacoes

# ------------------------------------------------------------------------------
# EXECUÇÃO DOS TESTES
# ------------------------------------------------------------------------------
if __name__ == "__main__":
    print("=== TESTE DA QUESTÃO 3 (Busca Binária Recursiva) ===")
    vetor_ordenado =
    alvo = 23
    indice = busca_binaria(vetor_ordenado, alvo)
    print(f"Vetor: {vetor_ordenado}")
    print(f"Alvo procurado: {alvo}")
    print(f"Índice retornado: {indice}\n")

    print("=== TESTE DA QUESTÃO 4 (Pior Caso com 1000 Elementos) ===")
    # Cria uma lista de 1000 números inteiros ordenados (1 a 1000)
    lista_1000 = list(range(1, 1001))
    # Número não presente na lista para forçar o pior caso
    alvo_inexistente = 1001

    _, comp_seq = busca_sequencial_com_contagem(lista_1000, alvo_inexistente)
    _, comp_bin = busca_binaria_com_contagem(lista_1000, alvo_inexistente)

    print(f"Tamanho da lista: {len(lista_1000)}")
    print(f"Alvo testado (inexistente): {alvo_inexistente}")
    print(f"Busca Sequencial - Total de comparações realizadas: {comp_seq}")
    print(f"Busca Binária    - Total de comparações realizadas: {comp_bin}")
```

#### **Resultado do Teste da Questão 4 (Pior Caso)**:
* **Busca Sequencial:** realiza **1000 comparações**.
* **Busca Binária:** realiza **10 comparações**.
````
