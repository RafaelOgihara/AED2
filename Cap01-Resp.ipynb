## Exercício 1
Criar a lista de frutas, adicionar `morango` ao final e imprimir.

## Exercício 2
Simular uma pilha. Como a pilha segue **LIFO**, o último livro inserido é o primeiro a sair.

```python
livros = ["O Pequeno Príncipe", "Dom Quixote", "1984"]
livro_removido = livros.pop()
print("Livro removido:", livro_removido)
print("Pilha restante:", livros)
```

## Exercício 3
Simular uma fila. Como a fila segue **FIFO**, o primeiro cliente é o primeiro a sair.

```python
clientes = ["Ana", "Bruno", "Carla"]
clientes.append("Daniel")
cliente_saiu = clientes.pop(0)
print("Cliente que saiu:", cliente_saiu)
print("Fila restante:", clientes)
```
