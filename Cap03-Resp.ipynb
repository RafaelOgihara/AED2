## Exercício 4
Função de busca sequencial para clientes.

```python
def buscar_cliente(nome, lista_clientes):
    for i in range(len(lista_clientes)):
        if lista_clientes[i] == nome:
            return i
    return -1

clientes = ["Ana", "Bruno", "Carla", "Daniel"]

indice = buscar_cliente("Carla", clientes)
if indice != -1:
    print("Cliente encontrado no índice", indice)
else:
    print("Cliente não encontrado.")

indice = buscar_cliente("Mariana", clientes)
if indice != -1:
    print("Cliente encontrado no índice", indice)
else:
    print("Cliente não encontrado.")
```

## Exercício 5
Contar quantas vezes um valor aparece na lista usando busca sequencial.

```python
def conta_ocorrencias(valor, lista):
    cont = 0
    for elemento in lista:
        if elemento == valor:
            cont += 1
    return cont

print(conta_ocorrencias(3, [1, 3, 5, 3, 7]))
print(conta_ocorrencias(2, [1, 2, 2, 4, 2]))
print(conta_ocorrencias(5, []))
```

## Exercício 6
Remover a primeira ocorrência de um valor e retornar o valor removido. Caso não encontre, retornar `None`.

```python
def remove_elemento(lista, valor):
    for i in range(len(lista)):
        if lista[i] == valor:
            return lista.pop(i)
    return None

numeros = [10, 20, 30, 20, 40]
removido = remove_elemento(numeros, 20)

print("Valor removido:", removido)
print("Lista restante:", numeros)
print("Valor removido:", remove_elemento(numeros, 99))
print("Lista vazia:", remove_elemento([], 10))
```
